````python
import datetime
````

````python
def complete_tasks():
    incomplete = {}
    current_day = datetime.datetime.now().strftime("%A") #current day of the week
    for list_day, list_task in tasks.items(): #looping through each day, and then each task to find incomplete tasks
        incomplete[list_day] = {}
        for list_task, completion in list_task.items():
            if completion == False: 
                incomplete[list_day][list_task] = False
            else:
                continue
    if len(incomplete) == 0:
        print("All tasks complete! Yay!") #if the there is nothing in incomplete, then there are no incomplete tasks
    else:
        print("Incomplete tasks are:", incomplete)
    completed_task = input("Enter completed task:").strip().title()
    completed_day = input("Enter day task completed:").strip().title()
    if completed_day not in tasks or completed_task not in tasks[completed_day]: #for tasks that were not included
        print("Day or task not found. Check that both are spelt correctly or originally included.")
        return None
    else:
        tasks[completed_day][completed_task] = True
        print("Task marked as complete.") #tasks marked as compelete
    if completed_day == current_day:
        print("And on time!") #for tasks completed that day
    elif day.index(completed_day) == day.index(current_day)-1:
        print("Sometimes it takes a day! That's okay!") #for tasks completed the day before 
    elif day.index(current_day) - day.index(completed_day) < 0:
        print("Are you a time traveller?")
    else:
        print("Better late than never.")
    return tasks
````

````python
def complete_habit():
    print(list(habits.keys())) #print just the habits
    completed_habit = input("Enter completed habit:").strip().title() #compeleted habit
    if completed_habit not in habits: #ensure that the task exists 
        print("Habit not found. Check that it is valid/spelt correctly.")
        return None
    print(habits[completed_habit])
    completed_date = input("Enter day habit was completed:").strip().title()
    if completed_date not in habits[completed_habit]: #ensure that the task date exists
        print("Day not found. Check that it is valid/spelt correctly.")
        return None
    habits[completed_habit][completed_date] = True
    print(f"Habit marked completed for the day {completed_date}. Keep on the journey!")
    return habits
````

### Removing Tasks and Habits

#### Instructions:
1. **Create a function to remove a task:**

- Make sure the day is valid. If it isn't, display a message saying the day is invalid.
- Show the list of tasks for that day. If there are none, display a message saying so.
- Ask the user which task to remove.
- If the task exists, delete it from the dictionary using del and display a message saying which task was deleted. If it doesn't exist, display a message saying so.

2. **Create a function to remove a habit:**
- If no habits are being tracked, display a message saying so.
- Display all habits that are currently being tracked.
- Ask the user which habit to remove.
- If it exists in the dictionary, delete it using del and display a message saying which habit was deleted. If it doesn't exist, display a message saying so.

#### Tips:
- Use **.strip()** and **.title()** to clean up user input.
- An empty dictionary counts as `False`, so `if not my_dict:` runs when the dictionary is empty. Use this to check if there's data before accessing or deleting something.
- `del` works with two sets of square brackets too. For example, `del pets['Rex']['Walk']` removes `'Walk'` from Rex's dictionary.
- **If the day, habit or task doesn't exist, show a helpful error message.**

````python
def del_task():
    day_task = input("Enter day of the week of task to delete:").strip().title() #the day that task will be deleted for
    if day_task not in day: #checks if the day exists and outputs an error message
        print("Not a valid day of the week.")
        return None
    pot_task = tasks[day_task] #potential tasks to be deleted as they are the tasks of the day
    if len(pot_task) == 0: #check if the dictionary for the day is empty
        print("No tasks set for that day.")
        return None
    else:
        print(pot_task)        
    task_tbd = input("Enter the task to delete:").strip().title() #task to be deleted
    if task_tbd not in pot_task: #check that the tasks exists
        print("Task not found for this day. Check to see if its in another day.")
    else:
        del pot_task[task_tbd] #delete the tasks 
        print("Task deleted!")
    return tasks
````

````python
def del_habit():
    if len(habits) == 0: #check to see if there are tracked habits
        print("No habits are currently being tracked.")
        return None
    else:
        print(habits)
    habit_tbd = input("Enter the habit to delete:").strip().title() #habit to be deleted
    if habit_tbd not in habits: #check that the habit exists
        print("Habit not found.")
    else:
        del habits[habit_tbd] #delete the habit 
        print("Habit deleted!")
    return habits
````

### Generating Weekly and Daily Reports

#### Instructions:
1. **Create a weekly report function**

This function should summarize the user's progress over the entire week.
- For habits:
    - If there are no habits, display a message saying so.
    - Loop through each habit in your habits dictionary.
    - Count how many days the habit was marked as complete, using a loop (for example, with a counter that starts at 0) or `sum()` (see the tips below).
    - Display the habit name and the number of days it was completed out of 7 (for example, "Drink Water: 3/7 days complete").
- For tasks:
    - Loop through all days and collect tasks into two separate lists:
        - One for completed tasks
        - One for not completed tasks
    - For each task, include the task name and its day in parentheses (for example, "Go to supermarket (Monday)").

2. **Create a daily report function**

This function allows the user to view all their activity for a specific day.

- Ask the user to enter a day (e.g., "Monday").
- If the day is valid:
    - Display all tasks for that day, showing whether each one is complete or not.
    - Display all habits, showing whether they were completed on that day.
    - If there are no tasks or no habits, display a message saying so.
- Use ✅ for completed and ❌ for not completed.

#### Tips:
- Use **.strip()** and **.title()** to make sure the input day matches the format in your list.
- Use clear formatting in your print statements to separate sections.
- Empty lists and dictionaries count as `False`, so `if not my_list:` runs when the list is empty.
- `sum()` can count how many items match a condition. For example, `sum(1 for t in [18, 25, 30] if t > 20)` gives `2` (it adds 1 for every temperature above 20).
- You can choose between two values in one line, for example: `label = "Yes" if bought else "No"`. Copy ✅ and ❌ from these instructions and paste them into your strings.
- **If the day doesn't exist, show a helpful error message.**

````python
def weekly_report():
    weekly_habit = [] #empty list for habits 
    if len(habits) == 0: #check to see if there are any tracked habits
        print("No habits tracked for this week.")
    else: 
        for w_habit, w_hday in habits.items(): #loop through habits 
            comp_rate = sum(1 for h_comp in w_hday.values() if h_comp == True) #completion rate is the amount of completed habits
            weekly_habit.append(f"{w_habit}: {comp_rate}/7 complete") #add to empty dictionary with "number/7 completed"
        print(f"Habits completion for this week are: {weekly_habit}")
    weekly_tasks_comp = [] #empty list for completed tasks
    weekly_tasks_incomp = [] #empty list for incompleted tasks
    if len(tasks) == 0: #check to see if there are any tracked tasks
        print("No tasks were tracked for this week.")
    else: 
        for w_tday, w_task in tasks.items(): #loop through the tasks and days 
            for w_task, w_comp in w_task.items(): #loop through the completion statements
                if w_comp == True:
                    weekly_tasks_comp.append(f"{w_task}({w_tday})") #completed tasks
                else:
                    weekly_tasks_incomp.append(f"{w_task}({w_tday})") #incompleted tasks
        if len(weekly_tasks_comp) == 0 and len(weekly_tasks_incomp) == 0: #an error as it would mean there are no tasks, or they're not marked as true/false
            print("Something has gone wrong! Try again!")
        elif len(weekly_tasks_comp) == 0:
            print(f"No completed tasks this week. The incompleted tasks were: {weekly_tasks_incomp}. Theres always next week!")
        elif len(weekly_tasks_incomp) == 0:
            print(f"Completed tasks were: {weekly_tasks_comp}. There are no incompleted tasks. Good job!")
        else: 
            print(f"Completed tasks this week were: {weekly_tasks_comp}. Good job! The incompleted tasks were: {weekly_tasks_incomp}. Theres always next week!")
    return 
````

````python
def daily_report():
    report_day = input("Enter day you would like a report for:").strip().title() #day for the report
    if report_day not in day: #checks if the day exists and outputs an error message
        print("Not a valid day of the week.")
        return None
    day_habits = []
    day_tasks = []
    for d_task, d_tcomp in tasks[report_day].items():
        d_t_comp = "✅" if d_tcomp == True else "❌"
        day_tasks.append(f"{d_task}({d_t_comp})")
    if len(day_tasks) == 0:
        print("For this day, there are no tasks listed.")
    else:
        print(f"Summary of the tasks for this day is {day_tasks}.")
    for d_habit in habits.keys(): #loop through habits but just the habits themselves 
        d_hcomp = habits[d_habit][report_day]
        d_h_comp = "✅" if d_hcomp == True else "❌"
        day_habits.append(f"{d_habit}({d_h_comp})")
    if len(day_habits) == 0:
        print("No habits created for this day.")
    else:
        print(f"The habits completion for this day is {day_habits}. Keep up the work!")
    return 
````
